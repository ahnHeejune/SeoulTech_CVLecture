## 1. 공부/연구에 도움을 줄만한 싸이트 

### 1) Kaggle 
- 데이터 사이언티스트들이 모여 1) 실력을 겨루고, 2) 데이터를 공유하며, 3) 함께 학습하는 데이터 분석 및 AI 커리큘럼 플랫폼 2010년에 설립되었으며, 현재는 구글(Google)의 자회사로 운영 중
- 아주 다양한 데이터와 코드들을 공개
- Google 계정등으로 쉽게 로그인 가능

### 2) 깃허브(GitHub)
- 전 세계 소프트웨어 개발자와 연구자들이 코드를 저장하고, 관리하며, 서로 협업하는 세계 최대의 오픈소스 코드 플랫폼. 
- 캐글이 '데이터셋과 분석 시합' 중심이라면, 깃허브는 '실제 작동하는 완성형 인공지능 프로그램과 연구 소스코드'가 중심이 되는 공간
- 무료 계정제공, public 및 private 세팅가능

### 3) 구글 코랩(Google Colab, Colaboratory)
- 구글에서 제공하는 웹 브라우저 기반의 인공지능/데이터 분석 실습 환경
- '설치 지옥'이 없습니다: 이미 모든 필수 라이브러리(PyTorch, Pandas, NumPy 등)가 설치 완료되어 있어 바로 코딩을 시작
- 무료 고성능 GPU 제공(T4 등)를 웹 서버에서 무료로 지원
- 구글 드라이브와 완벽한 연동: 연구에 사용할 캐글 데이터셋이나 설문 결과 CSV 파일을 본인의 구글 드라이브에 올려두고, 코드 한 줄로 쉽게 불러와 분석.

### 다음 코드의 Collab 실습 예

1. Collab 싸이트로 들어 간다. https://colab.research.google.com/
2. 메뉴의 [파일] -> [노트 열기]에서 [GitHub] 탭을 클릭합니다.
3. 검색창에 원하는 GitHub 저장소(Repository) 주소 (https://github.com/ahnHeejune/SeoulTech_CVLecture.git) 를 붙여넣고 오른쪽의 돋보기 버튼(검색)을 누릅니다
4. 아래에 해당 저장소에 들어있는 .ipynb 파일 목록이 나타납니다. 실행하고 싶은 파일 이름을 클릭합니다.
5. 주피터 노트북 파일이 코랩 화면에 그대로 열립니다.
6. 이제 상단부터 차례대로 재생 버튼(▶)을 누르며 실행하시면 됩니다.

## 2. 학습 조언 

- 기본 이론 이외에 이를 활용하기 위한 Python 패키지 사용법을 하나씩 일혀야 한다.
- 이분야에서 기본적으로 사용하는 package

   - numpy : Python에서 배별데이터 처리하는 가장 핵심적인 기능을 제공
   - padas : 데이터 파일기반  데이터 처리 기능을 제공
   - scikit-learn:  고전적, 현대적 ML기법 등을 지원
   - Pytorch (Tensorflow) : 신경망 처리를 지원하는 라이브러리
        
- "공부에는 왕도가 없다"라는 격억과 같이 너무 결과를 빨리 볼 생각을 하지 말고 이러한 툴을 사용하여 데이터를 직접 이해하는 시간을 충분히 갖는 것이 필요하다.
